# UNESCO Chat Analysis - February to May 2025

This analysis focuses on chat data from February to May 2025, including:
- Total number of sessions
- Total number of messages
- Message statistics per chat
- Conversation durations

```python
# Read CSV file and print the first 5 rows

import pandas as pd

# Read the CSV file
df = pd.read_csv('data.csv', sep=';')

# Convert the date column to datetime right after reading the file
df['m.creation_date'] = pd.to_datetime(df['m.creation_date'])

# Filter data between February 2025 and May 2025
start_date = '2025-02-01'
end_date = '2025-05-31'
df_filtered = df[(df['m.creation_date'] >= start_date) & (df['m.creation_date'] <= end_date)]

# Group by m.session and show the result
chats = df_filtered.groupby('m.session').size().reset_index(name='messages')

# Count the number of chats
chats_count = chats['m.session'].count()
print(f'Number of chats (fev-mai/2025): {chats_count}')

# Count the number of messages
messages_count = chats['messages'].sum()
print(f'Number of mensagens (fev-mai/2025): {messages_count}')

# Average number of messages per chat
average_messages = messages_count / chats_count
print(f'Average per chat: {average_messages:.2f}')

# Maximum and minimum messages in a chat
max_messages = chats['messages'].max()
min_messages = chats['messages'].min()
print(f'Max per chat: {max_messages}')
print(f'Min per chat: {min_messages}')

# Calculate chat durations
chat_durations = df_filtered.groupby('m.session').agg({
    'm.creation_date': lambda x: (x.max() - x.min()).total_seconds() / 60  # Convertendo para minutos
}).reset_index()

# Duration statistics
mean_duration = chat_durations['m.creation_date'].mean()
max_duration = chat_durations['m.creation_date'].max()
min_duration = chat_durations['m.creation_date'].min()

print('\nChat duration statistics:')
print(f'Average: {mean_duration:.2f} minutes')
print(f'Max: {max_duration:.2f} minutes')
print(f'Min: {min_duration * 60:.2f} seconds')
```
