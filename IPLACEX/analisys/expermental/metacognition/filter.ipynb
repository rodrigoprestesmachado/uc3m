```python
import pandas as pd

directory = 'gpt4o/'

# Fix: First parse the CSV and then split the combined column into separate columns
df = pd.read_csv(directory + 'result.csv', encoding='utf-8', sep=';', on_bad_lines='skip')

# Split the combined column into separate columns
if 'message_id,student_query,Metacognition' in df.columns:
	# Split the single column into three separate columns
	df = df['message_id,student_query,Metacognition'].str.split(',', n=2, expand=True)
	df.columns = ['message_id', 'student_query', 'Metacognition']
	# Convert Metacognition to integer
	df['Metacognition'] = pd.to_numeric(df['Metacognition'], errors='coerce').fillna(0).astype(int)

# Now filter for metacognition=1
df_metacognition = df[df['Metacognition'] == 1]

df_metacognition.to_csv(directory + 'phrases.csv', index=False, encoding='utf-8', sep=';')

df_metacognition.head(30)



```
