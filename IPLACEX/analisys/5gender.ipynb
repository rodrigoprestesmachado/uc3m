```python
import pandas as pd

# Reading the CSV from the current directory
df = pd.read_csv('data/temp/chats_edu.csv')

# Filter sessions with more than 4 messages
df = df[df['session_id'].map(df['session_id'].value_counts()) > 2 ]

# Count the number of sessions
session_count = df['session_id'].nunique()
# Print the number of sessions with more than 4 messages
print(f"Sessions with more than 4 messages: {session_count}")

# agrupar por session_id
grouped_df = df[['session_id', 'user_id', 'SEXO', 'FECNAC']].drop_duplicates().reset_index(drop=True)

# quero saber quem são esses usuarios
# Criar um novo DataFrame apenas com user_id únicos
unique_users_df = grouped_df[['user_id', 'SEXO', 'FECNAC']].drop_duplicates().reset_index(drop=True)
unique_users_df['ANO_NASC'] = pd.to_datetime('1899-12-30') + pd.to_timedelta(unique_users_df['FECNAC'], unit='D')
unique_users_df['ANO_NASC'] = unique_users_df['ANO_NASC'].dt.year
unique_users_df.head(220)

# contar o número de usuários únicos
unique_user_count = unique_users_df['user_id'].nunique()
# Print the number of unique users
print(f"Number of unique users: {unique_user_count}")

# Contas quantos são masculinos e quantos são femininos
unique_user_count_sexo = unique_users_df['SEXO'].value_counts()
print(unique_user_count_sexo)



```

```python
import pandas as pd
import numpy as np
from scipy.stats import chi2_contingency


# Dados observados: linhas = tipo de chat, colunas = gênero
dados = pd.DataFrame({
    'Feminino': [48, 245],
    'Masculino': [31, 322]
}, index=['Mais de 1 mensagen', '1 mensagem'])

# Teste do qui-quadrado de independência
chi2, p, dof, expected = chi2_contingency(dados)

# Mostrar resultados
print(f"Estatística qui-quadrado: {chi2:.4f}")
print(f"p-valor: {p:.6f}")
print(f"Graus de liberdade: {dof}")
print("Valores esperados:")
print(pd.DataFrame(expected, index=dados.index, columns=dados.columns))
```
