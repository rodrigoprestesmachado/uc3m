## Delta Calculous

```python
import pandas as pd

# Read the CSV file with the correct separator
df = pd.read_csv('data/iplacex.csv', encoding='utf-8', sep=';')

# If needed, process the data to calculate deltas
if 'session_id' in df.columns and 'timestamp' in df.columns:
    # Sort the DataFrame by session_id and timestamp
    df_sorted = df.sort_values(['session_id', 'timestamp'])

    # Convert Excel timestamp (replace comma with dot and convert from Excel days to datetime)
    df_sorted['timestamp'] = df_sorted['timestamp'].str.replace(',', '.')
    df_sorted['timestamp'] = pd.to_datetime(pd.to_numeric(df_sorted['timestamp'], errors='coerce'), unit='D', origin='1899-12-30')

    # Calculate deltas
    df_sorted['delta'] = df_sorted['timestamp'].diff().dt.total_seconds() / 60
    df_sorted['delta'] = df_sorted.apply(lambda row:
        0 if pd.isna(row['delta']) or df_sorted['session_id'].shift(1).loc[row.name] != row['session_id']
        else row['delta'], axis=1)

    # Round and save
    df_sorted['delta'] = df_sorted['delta'].round(2)
    df_sorted.to_csv('data/temp/chats_delta.csv', index=False, encoding='utf-8')
    print("Deltas calculated and saved to 'data/temp/chats_delta.csv'")
else:
    print("The required columns 'session_id' and 'timestamp' are not present in the DataFrame.")
    print("Available columns:", df.columns)
    print("Please check the CSV file format and ensure it contains the necessary data.")
```
