# External Spacing

```python
import pandas as pd

# Load CSV file
df = pd.read_csv('data/iplacex.csv')

# Group data by user and date, counting unique sessions
sessions_by_user = df.groupby(['user_id', 'date'])['session_id'].nunique().reset_index()

# Rename columns for better understanding
sessions_by_user.columns = ['User', 'Date', 'Number of Sessions']

# Sort by user
sessions_by_user = sessions_by_user.sort_values(['User'])

# Group by user and calculates de mean
mean_sessions = sessions_by_user.groupby('User')['Number of Sessions'].mean().reset_index()

# Total os users
total_users = mean_sessions['User'].count()
print(f'Total number of users: {total_users}')

# Calculate the mean number of sessions per user
mean_sessions_day = mean_sessions['Number of Sessions'].mean().round(2)
print(f'Mean number of sessions per user in which day: {mean_sessions_day}')

# save the result to a new CSV file
mean_sessions.to_csv('data/temp/mean_sessions_by_user.csv', index=False)
```
